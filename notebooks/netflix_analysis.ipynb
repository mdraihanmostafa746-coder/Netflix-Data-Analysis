```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="%B %d, %Y",
    errors="coerce"
)

df["day_added"] = df["date_added"].dt.day
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
```

```python
ax = sns.countplot(
    x='type',
    data=df,
    order=df['type'].value_counts().index,
    hue='type',
    palette={'Movie': 'red', 'TV Show': 'orange'},
    legend=False
)

for container in ax.containers:
    ax.bar_label(container)
plt.title("Movies vs TV Shows on Netflix")
```

```python
x = sns.countplot(
    x='rating',
    data=df,
    order=df['rating'].value_counts().index,
    hue='rating',
    palette={
        'TV-MA': '#E50914',
        'TV-14': '#2563EB',
        'R': '#0D9488',
        'PG-13': '#F59E0B',
        'TV-PG': '#7C3AED',
        'PG': '#6B7280',
        'TV-G': '#F97316',
        'TV-Y7': '#38BDF8',
        'TV-Y': '#22C55E',
        'NR': '#14B8A6',
        'G': '#FBBF24',
        'TV-Y7-FV': '#EF4444',
        'UR': '#FB923C',
        'NC-17': '#6D28D9'
    },
    legend=False
)

for container in x.containers:
    x.bar_label(container)

plt.title("Distribution of Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 8))

sns.countplot(
    x='rating',
    hue='type',
    data=df,
    palette={
        'Movie': '#E50914',      
        'TV Show': '#2563EB'  
    }
)

plt.title('Relation between Type and Rating')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=90)

plt.show()
```
